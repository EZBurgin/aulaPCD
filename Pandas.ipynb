```python
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/andr3nf/clima_municipios_brasileiros/refs/heads/main/clima_historico_municipal.csv')

df_tucunduva = df[df['nome_municipio'].str.contains('Tucunduva', case=False, na=False)]

df_tucunduva
```

```python
# 1. Mês de temperatura mínima
monthly_temp_columns = [col for col in df_tucunduva.columns if 'temp_media_' in col]
min_temp_month = df_tucunduva[monthly_temp_columns].idxmin(axis=1).iloc[0]
min_temp_value = df_tucunduva[monthly_temp_columns].min(axis=1).iloc[0]
print(f"Mês com a temperatura média mínima: {min_temp_month.replace('temp_media_', '').capitalize()} ({min_temp_value:.2f}°C)")
```

```python
# 2. Mês de maior déficit hídrico
monthly_def_columns = [col for col in df_tucunduva.columns if 'def_' in col]
max_def_month = df_tucunduva[monthly_def_columns].idxmax(axis=1).iloc[0]
max_def_value = df_tucunduva[monthly_def_columns].max(axis=1).iloc[0]
print(f"Mês com o maior déficit hídrico: {max_def_month.replace('def_', '').capitalize()} ({max_def_value:.2f})")
```

```python
# 3. Média de temperatura pelas 4 estações do ano
# Definir as colunas de temperatura para cada estação
summer_months = ['temp_media_dez', 'temp_media_jan', 'temp_media_fev']
autumn_months = ['temp_media_mar', 'temp_media_abr', 'temp_media_mai']
winter_months = ['temp_media_jun', 'temp_media_jul', 'temp_media_ago']
spring_months = ['temp_media_set', 'temp_media_out', 'temp_media_nov']

# Calcular a média para cada estação
summer_avg_temp = df_tucunduva[summer_months].mean(axis=1).iloc[0]
autumn_avg_temp = df_tucunduva[autumn_months].mean(axis=1).iloc[0]
winter_avg_temp = df_tucunduva[winter_months].mean(axis=1).iloc[0]
spring_avg_temp = df_tucunduva[spring_months].mean(axis=1).iloc[0]

print(f"Média de temperatura no Verão: {summer_avg_temp:.2f}°C")
print(f"Média de temperatura no Outono: {autumn_avg_temp:.2f}°C")
print(f"Média de temperatura no Inverno: {winter_avg_temp:.2f}°C")
print(f"Média de temperatura na Primavera: {spring_avg_temp:.2f}°C")
```
