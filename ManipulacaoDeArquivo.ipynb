```python
import json

idade = 20
nome = "Eduardo"
altura = 1.87

personagem_dic = {
    "nome": nome,
    "idade": idade,
    "altura": altura
}

print(personagem_dic)

with open('aulaPCD/personagem.txt', 'w') as json_file:
  json.dump(personagem_dic, json_file, indent=4)

with open('aulaPCD/personagem.txt', 'r') as json_file:
  personagem_arquivo = json.load(json_file)

print(json.dumps(personagem_arquivo, indent=4))

```
